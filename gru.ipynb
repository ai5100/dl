```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense

uploaded = files.upload()

file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)

print("\nCSV Columns:")
print(df.columns.tolist())

column = input("\nEnter the traffic column name: ")

data = df[column].dropna().values.reshape(-1, 1)

scaler = MinMaxScaler()
data_scaled = scaler.fit_transform(data)

time_steps = int(input("Enter number of previous time steps: "))

X = []
y = []

for i in range(time_steps, len(data_scaled)):
    X.append(data_scaled[i-time_steps:i])
    y.append(data_scaled[i])

X = np.array(X)
y = np.array(y)

print("\nInput Shape:", X.shape)
print("Output Shape:", y.shape)

model = Sequential([
    GRU(64, return_sequences=True, input_shape=(time_steps, 1)),
    GRU(32),
    Dense(16, activation='relu'),
    Dense(1)
])

model.compile(
    optimizer='adam',
    loss='mse'
)

epochs = int(input("\nEnter number of epochs: "))

model.fit(
    X,
    y,
    epochs=epochs,
    batch_size=8,
    verbose=1
)

predicted = model.predict(X)

predicted = scaler.inverse_transform(predicted)
actual = scaler.inverse_transform(y)

print("\nActual vs Predicted Traffic")

for i in range(len(actual)):
    print(
        f"Actual: {actual[i][0]:.2f} "
        f"| Predicted: {predicted[i][0]:.2f}"
    )

last_sequence = data_scaled[-time_steps:]
last_sequence = last_sequence.reshape(1, time_steps, 1)

next_prediction = model.predict(last_sequence)
next_prediction = scaler.inverse_transform(next_prediction)

print(
    "\nPredicted Traffic for Next Time Step:",
    round(float(next_prediction[0][0]), 2)
)

plt.figure(figsize=(10, 5))
plt.plot(actual, label="Actual Traffic")
plt.plot(predicted, label="Predicted Traffic")
plt.xlabel("Time")
plt.ylabel("Traffic Volume")
plt.title("GRU-Based Traffic Flow Prediction")
plt.legend()
plt.show()

WASTE GENERATION:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense

uploaded = files.upload()

file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)

print("\nCSV Columns:")
print(df.columns.tolist())

column = input("\nEnter the waste column name: ")

data = df[column].dropna().values.reshape(-1, 1)

scaler = MinMaxScaler()
data_scaled = scaler.fit_transform(data)

time_steps = int(input("Enter number of previous time steps: "))

X = []
y = []

for i in range(time_steps, len(data_scaled)):
    X.append(data_scaled[i-time_steps:i])
    y.append(data_scaled[i])

X = np.array(X)
y = np.array(y)

print("\nInput Shape:", X.shape)
print("Output Shape:", y.shape)

model = Sequential([
    GRU(64, return_sequences=True, input_shape=(time_steps, 1)),
    GRU(32),
    Dense(16, activation='relu'),
    Dense(1)
])

model.compile(
    optimizer='adam',
    loss='mse'
)

epochs = int(input("\nEnter number of epochs: "))

model.fit(
    X,
    y,
    epochs=epochs,
    batch_size=8,
    verbose=1
)

predicted = model.predict(X)

predicted = scaler.inverse_transform(predicted)
actual = scaler.inverse_transform(y)

print("\nActual vs Predicted Waste")

for i in range(len(actual)):
    print(
        f"Actual: {actual[i][0]:.2f} "
        f"| Predicted: {predicted[i][0]:.2f}"
    )

last_sequence = data_scaled[-time_steps:]
last_sequence = last_sequence.reshape(1, time_steps, 1)

next_prediction = model.predict(last_sequence)
next_prediction = scaler.inverse_transform(next_prediction)

print(
    "\nPredicted Waste for Next Time Step:",
    round(float(next_prediction[0][0]), 2)
)

plt.figure(figsize=(10, 5))
plt.plot(actual, label="Actual Waste")
plt.plot(predicted, label="Predicted Waste")
plt.xlabel("Time")
plt.ylabel("Waste Quantity")
plt.title("GRU-Based Waste Generation Prediction")
plt.legend()
plt.show()
```
