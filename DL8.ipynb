```python
SINGLE DIGIT:
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from PIL import Image, ImageOps

(x_train, y_train), (x_test, y_test) = datasets.mnist.load_data()

print("Training images :", x_train.shape)
print("Testing images  :", x_test.shape)

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

model = models.Sequential([
    layers.Conv2D(32, (3, 3), activation="relu", input_shape=(28, 28, 1)),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(10, activation="softmax")
])

model.summary()

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history = model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.1
)

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("\n========== FINAL RESULTS ==========")
print("Training Images :", len(x_train))
print("Testing Images  :", len(x_test))
print("Image Size      : 28 x 28")
print("Number of Classes :", 10)
print("Test Loss       :", test_loss)
print("Test Accuracy   : {:.2f}%".format(test_accuracy * 100))

plt.figure(figsize=(8, 5))
plt.plot(history.history["accuracy"], marker="o", label="Training Accuracy")
plt.plot(history.history["val_accuracy"], marker="o", label="Validation Accuracy")
plt.title("CNN Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(history.history["loss"], marker="o", label="Training Loss")
plt.plot(history.history["val_loss"], marker="o", label="Validation Loss")
plt.title("CNN Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()

print("\n======================================")
print("       DIGIT RECOGNITION SYSTEM")
print("======================================")

uploaded = files.upload()
file_name = list(uploaded.keys())[0]

print("Input Image :", file_name)

image = Image.open(file_name).convert("L")
image = image.resize((28, 28))

image_array = np.array(image)

if np.mean(image_array) > 127:
    image_array = np.array(ImageOps.invert(Image.fromarray(image_array)))

image_array = image_array.astype("float32") / 255.0
image_array = image_array.reshape(1, 28, 28, 1)

prediction = model.predict(image_array, verbose=0)

predicted_digit = np.argmax(prediction[0])
confidence = prediction[0][predicted_digit] * 100

print("\n========== PREDICTION ==========")
print("Input Image     :", file_name)
print("Predicted Digit :", predicted_digit)
print("Confidence      : {:.2f}%".format(confidence))
print("================================")

plt.figure(figsize=(4, 4))
plt.imshow(image_array.reshape(28, 28), cmap="gray")
plt.title(
    "Predicted Digit: {} ({:.2f}%)".format(
        predicted_digit,
        confidence
    )
)
plt.axis("off")
plt.show()

MULTIDIGIT:
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from PIL import Image, ImageOps
import cv2

(x_train, y_train), (x_test, y_test) = datasets.mnist.load_data()

print("Training images :", x_train.shape)
print("Testing images  :", x_test.shape)

x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)

model = models.Sequential([
    layers.Input(shape=(28, 28, 1)),
    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(128, activation="relu"),
    layers.Dense(10, activation="softmax")
])

model.summary()

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history = model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.1
)

test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test,
    verbose=0
)

print("\n========== FINAL RESULTS ==========")
print("Training Images :", len(x_train))
print("Testing Images  :", len(x_test))
print("Image Size      : 28 x 28")
print("Number of Classes :", 10)
print("Test Loss       :", test_loss)
print("Test Accuracy   : {:.2f}%".format(test_accuracy * 100))

plt.figure(figsize=(8, 5))
plt.plot(
    history.history["accuracy"],
    marker="o",
    label="Training Accuracy"
)
plt.plot(
    history.history["val_accuracy"],
    marker="o",
    label="Validation Accuracy"
)
plt.title("CNN Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(
    history.history["loss"],
    marker="o",
    label="Training Loss"
)
plt.plot(
    history.history["val_loss"],
    marker="o",
    label="Validation Loss"
)
plt.title("CNN Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()

print("\n======================================")
print("       MULTI-DIGIT RECOGNITION")
print("======================================")

uploaded = files.upload()
file_name = list(uploaded.keys())[0]

print("Input Image :", file_name)

image = Image.open(file_name).convert("L")
image_array = np.array(image)

if np.mean(image_array) > 127:
    image_array = np.array(
        ImageOps.invert(
            Image.fromarray(image_array)
        )
    )

_, binary = cv2.threshold(
    image_array,
    0,
    255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

contours, _ = cv2.findContours(
    binary,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

digit_boxes = []

for contour in contours:
    x, y, w, h = cv2.boundingRect(contour)

    if w >= 5 and h >= 10:
        digit_boxes.append((x, y, w, h))

digit_boxes = sorted(
    digit_boxes,
    key=lambda box: box[0]
)

print(
    "Number of digits detected :",
    len(digit_boxes)
)

predicted_digits = []
confidence_values = []
digit_images = []

for x, y, w, h in digit_boxes:

    digit = binary[y:y+h, x:x+w]

    padding = 10

    digit = cv2.copyMakeBorder(
        digit,
        padding,
        padding,
        padding,
        padding,
        cv2.BORDER_CONSTANT,
        value=0
    )

    digit = cv2.resize(
        digit,
        (28, 28),
        interpolation=cv2.INTER_AREA
    )

    digit = digit.astype("float32") / 255.0

    digit_input = digit.reshape(
        1, 28, 28, 1
    )

    prediction = model.predict(
        digit_input,
        verbose=0
    )

    predicted_digit = np.argmax(
        prediction[0]
    )

    confidence = (
        prediction[0][predicted_digit] * 100
    )

    predicted_digits.append(
        predicted_digit
    )

    confidence_values.append(
        confidence
    )

    digit_images.append(digit)

predicted_number = "".join(
    str(digit)
    for digit in predicted_digits
)

print("\n========== PREDICTION ==========")
print("Input Image      :", file_name)
print("Detected Digits  :", len(predicted_digits))
print("Predicted Number :", predicted_number)

print("\nIndividual Predictions:")

for i in range(len(predicted_digits)):
    print(
        "Digit",
        i + 1,
        ":",
        predicted_digits[i],
        "Confidence:",
        "{:.2f}%".format(
            confidence_values[i]
        )
    )

print("================================")

if len(digit_images) > 0:

    plt.figure(
        figsize=(3 * len(digit_images), 4)
    )

    for i in range(len(digit_images)):

        plt.subplot(
            1,
            len(digit_images),
            i + 1
        )

        plt.imshow(
            digit_images[i],
            cmap="gray"
        )

        plt.title(
            "{}\n{:.2f}%".format(
                predicted_digits[i],
                confidence_values[i]
            )
        )

        plt.axis("off")

    plt.suptitle(
        "Detected Digits: " + predicted_number
    )

    plt.tight_layout()
    plt.show()
```
