```python
SGD:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


filename = input("Enter CSV file name: ")
data = pd.read_csv(filename)

X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values.reshape(-1, 1)


X = (X - X.mean(axis=0)) / X.std(axis=0)


learning_rate = 0.01
epochs = 500

n_samples, n_features = X.shape

np.random.seed(42)

weights = np.random.randn(n_features, 1)
bias = np.random.randn()

losses = []


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


for epoch in range(epochs):

    indices = np.random.permutation(n_samples)
    X_shuffled = X[indices]
    y_shuffled = y[indices]

    total_loss = 0

    for i in range(n_samples):

        xi = X_shuffled[i].reshape(1, -1)
        yi = y_shuffled[i]

        z = np.dot(xi, weights) + bias
        output = sigmoid(z)

        error = yi - output

        loss = error ** 2
        total_loss += loss

        gradient = error * output * (1 - output)

        weights += learning_rate * gradient * xi.T
        bias += learning_rate * gradient

    losses.append(total_loss[0][0] / n_samples)


predictions = sigmoid(np.dot(X, weights) + bias)
predictions = (predictions >= 0.5).astype(int)

accuracy = np.mean(predictions == y) * 100


print("\n==============================")
print("Training Completed Successfully")
print("==============================")

print("\nWeights:")
print(weights)

print("\nBias:")
print(bias)

print("\nFinal Loss = {:.6f}".format(losses[-1]))
print("Accuracy = {:.2f}%".format(accuracy))


plt.figure(figsize=(8,5))
plt.plot(losses)

plt.title("SGD Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.show()

ADAGRAD:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

filename = input("Enter CSV file name: ")
data = pd.read_csv(filename)

X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values.reshape(-1, 1)


X = (X - X.mean(axis=0)) / X.std(axis=0)


learning_rate = 0.1
epochs = 200
epsilon = 1e-8

n_samples, n_features = X.shape


np.random.seed(42)

weights = np.random.randn(n_features, 1)
bias = np.random.randn()

G_w = np.zeros((n_features, 1))
G_b = 0

losses = []


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


for epoch in range(epochs):

    indices = np.random.permutation(n_samples)
    X_shuffled = X[indices]
    y_shuffled = y[indices]

    total_loss = 0

    for i in range(n_samples):

        xi = X_shuffled[i].reshape(1, -1)
        yi = y_shuffled[i]

        z = np.dot(xi, weights) + bias
        output = sigmoid(z)

        error = yi - output

        loss = error ** 2
        total_loss += loss

        grad_w = -2 * error * output * (1 - output) * xi.T
        grad_b = -2 * error * output * (1 - output)

        G_w += grad_w ** 2
        G_b += grad_b ** 2

        weights -= (learning_rate / (np.sqrt(G_w) + epsilon)) * grad_w
        bias -= (learning_rate / (np.sqrt(G_b) + epsilon)) * grad_b

    losses.append(total_loss[0][0] / n_samples)


pred = sigmoid(np.dot(X, weights) + bias)
pred = (pred >= 0.5).astype(int)

accuracy = np.mean(pred == y) * 100

print("\n==============================")
print("Training Completed Successfully")
print("==============================")

print("\nWeights")
print(weights)

print("\nBias")
print(bias)

print("\nFinal Loss = {:.6f}".format(losses[-1]))
print("Accuracy = {:.2f}%".format(accuracy))

plt.figure(figsize=(8,5))
plt.plot(losses)
plt.title("AdaGrad - Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.show()

MINIBATCH:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

filename = input("Enter CSV file name: ")
data = pd.read_csv(filename)

X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values.reshape(-1, 1)

X = (X - X.mean(axis=0)) / X.std(axis=0)

learning_rate = 0.01
epochs = 500
batch_size = 16

n_samples, n_features = X.shape

np.random.seed(42)
weights = np.random.randn(n_features, 1)
bias = np.random.randn()

losses = []

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

for epoch in range(epochs):

    indices = np.random.permutation(n_samples)
    X_shuffled = X[indices]
    y_shuffled = y[indices]

    total_loss = 0

    for i in range(0, n_samples, batch_size):

        xi = X_shuffled[i:i+batch_size]
        yi = y_shuffled[i:i+batch_size]

        z = np.dot(xi, weights) + bias
        output = sigmoid(z)

        error = yi - output

        loss = np.mean(error ** 2)
        total_loss += loss * len(xi)

        gradient = error * output * (1 - output)

        weights += learning_rate * np.dot(xi.T, gradient) / len(xi)
        bias += learning_rate * np.mean(gradient)

    losses.append(total_loss / n_samples)

predictions = sigmoid(np.dot(X, weights) + bias)
predictions = (predictions >= 0.5).astype(int)

accuracy = np.mean(predictions == y) * 100

print("\nTraining Completed Successfully")
print("\nWeights:")
print(weights)
print("\nBias:")
print(bias)
print("\nFinal Loss = {:.6f}".format(losses[-1]))
print("Accuracy = {:.2f}%".format(accuracy))

plt.figure(figsize=(8,5))
plt.plot(losses)
plt.title("Mini-Batch Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.show()

MOMENTUM GRADIENT:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

filename = input("Enter CSV file name: ")
data = pd.read_csv(filename)

X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values.reshape(-1, 1)

X = (X - X.mean(axis=0)) / X.std(axis=0)

learning_rate = 0.01
epochs = 500
momentum = 0.9

n_samples, n_features = X.shape

np.random.seed(42)
weights = np.random.randn(n_features, 1)
bias = np.random.randn()

v_w = np.zeros_like(weights)
v_b = 0

losses = []

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

for epoch in range(epochs):

    indices = np.random.permutation(n_samples)
    X_shuffled = X[indices]
    y_shuffled = y[indices]

    total_loss = 0

    for i in range(n_samples):

        xi = X_shuffled[i].reshape(1, -1)
        yi = y_shuffled[i]

        z = np.dot(xi, weights) + bias
        output = sigmoid(z)

        error = yi - output

        loss = error ** 2
        total_loss += loss

        gradient = error * output * (1 - output)

        dw = -gradient * xi.T
        db = -gradient

        v_w = momentum * v_w - learning_rate * dw
        v_b = momentum * v_b - learning_rate * db

        weights += v_w
        bias += v_b

    losses.append(total_loss[0][0] / n_samples)

predictions = sigmoid(np.dot(X, weights) + bias)
predictions = (predictions >= 0.5).astype(int)

accuracy = np.mean(predictions == y) * 100

print("\nTraining Completed Successfully")
print("\nWeights:")
print(weights)
print("\nBias:")
print(bias)
print("\nFinal Loss = {:.6f}".format(losses[-1]))
print("Accuracy = {:.2f}%".format(accuracy))

plt.figure(figsize=(8,5))
plt.plot(losses)
plt.title("Momentum Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.show()


NAG:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

filename = input("Enter CSV file name: ")
data = pd.read_csv(filename)

X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values.reshape(-1, 1)

X = (X - X.mean(axis=0)) / X.std(axis=0)

learning_rate = 0.01
epochs = 500
momentum = 0.9

n_samples, n_features = X.shape

np.random.seed(42)
weights = np.random.randn(n_features, 1)
bias = np.random.randn()

v_w = np.zeros_like(weights)
v_b = 0

losses = []

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

for epoch in range(epochs):

    indices = np.random.permutation(n_samples)
    X_shuffled = X[indices]
    y_shuffled = y[indices]

    total_loss = 0

    for i in range(n_samples):

        xi = X_shuffled[i].reshape(1, -1)
        yi = y_shuffled[i]

        look_w = weights + momentum * v_w
        look_b = bias + momentum * v_b

        z = np.dot(xi, look_w) + look_b
        output = sigmoid(z)

        error = yi - output

        loss = error ** 2
        total_loss += loss

        gradient = error * output * (1 - output)

        dw = -gradient * xi.T
        db = -gradient

        v_w = momentum * v_w - learning_rate * dw
        v_b = momentum * v_b - learning_rate * db

        weights += v_w
        bias += v_b

    losses.append(total_loss[0][0] / n_samples)

predictions = sigmoid(np.dot(X, weights) + bias)
predictions = (predictions >= 0.5).astype(int)

accuracy = np.mean(predictions == y) * 100

print("\nTraining Completed Successfully")
print("\nWeights:")
print(weights)
print("\nBias:")
print(bias)
print("\nFinal Loss = {:.6f}".format(losses[-1]))
print("Accuracy = {:.2f}%".format(accuracy))

plt.figure(figsize=(8,5))
plt.plot(losses)
plt.title("Nesterov Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.show()

ADAM:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

filename = input("Enter CSV file name: ")
data = pd.read_csv(filename)

X = data.iloc[:, :-1].values
y = data.iloc[:, -1].values.reshape(-1, 1)

X = (X - X.mean(axis=0)) / X.std(axis=0)

learning_rate = 0.01
epochs = 500

beta1 = 0.9
beta2 = 0.999
epsilon = 1e-8

n_samples, n_features = X.shape

np.random.seed(42)
weights = np.random.randn(n_features, 1)
bias = np.random.randn()

m_w = np.zeros_like(weights)
v_w = np.zeros_like(weights)

m_b = 0
v_b = 0

t = 0
losses = []

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

for epoch in range(epochs):

    indices = np.random.permutation(n_samples)
    X_shuffled = X[indices]
    y_shuffled = y[indices]

    total_loss = 0

    for i in range(n_samples):

        xi = X_shuffled[i].reshape(1, -1)
        yi = y_shuffled[i]

        z = np.dot(xi, weights) + bias
        output = sigmoid(z)

        error = yi - output

        loss = error ** 2
        total_loss += loss

        gradient = error * output * (1 - output)

        dw = -gradient * xi.T
        db = -gradient

        t += 1

        m_w = beta1 * m_w + (1 - beta1) * dw
        v_w = beta2 * v_w + (1 - beta2) * (dw ** 2)

        m_b = beta1 * m_b + (1 - beta1) * db
        v_b = beta2 * v_b + (1 - beta2) * (db ** 2)

        m_w_corrected = m_w / (1 - beta1 ** t)
        v_w_corrected = v_w / (1 - beta2 ** t)

        m_b_corrected = m_b / (1 - beta1 ** t)
        v_b_corrected = v_b / (1 - beta2 ** t)

        weights -= learning_rate * m_w_corrected / (np.sqrt(v_w_corrected) + epsilon)
        bias -= learning_rate * m_b_corrected / (np.sqrt(v_b_corrected) + epsilon)

    losses.append(total_loss[0][0] / n_samples)

predictions = sigmoid(np.dot(X, weights) + bias)
predictions = (predictions >= 0.5).astype(int)

accuracy = np.mean(predictions == y) * 100

print("\nTraining Completed Successfully")
print("\nWeights:")
print(weights)
print("\nBias:")
print(bias)
print("\nFinal Loss = {:.6f}".format(losses[-1]))
print("Accuracy = {:.2f}%".format(accuracy))

plt.figure(figsize=(8,5))
plt.plot(losses)
plt.title("Adam Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True)
plt.show()
```
