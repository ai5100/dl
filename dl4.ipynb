```python

import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

def mse_loss(y, output):
    return np.mean((y - output) ** 2)


print("DNN with 2 Hidden Layers")
print("-------------------------")

x1 = float(input("Enter input X1: "))
x2 = float(input("Enter input X2: "))

X = np.array([[x1, x2]])

target = float(input("Enter target output: "))
Y = np.array([[target]])

lr = float(input("Enter learning rate: "))
epochs = int(input("Enter number of epochs: "))


print("\nEnter weights for Hidden Layer 1")

w11 = float(input("Weight X1 -> H1: "))
w12 = float(input("Weight X2 -> H1: "))
w21 = float(input("Weight X1 -> H2: "))
w22 = float(input("Weight X2 -> H2: "))

W1 = np.array([
    [w11, w21],
    [w12, w22]
])


print("\nEnter biases for Hidden Layer 1")

b1_1 = float(input("Bias H1: "))
b1_2 = float(input("Bias H2: "))

b1 = np.array([[b1_1, b1_2]])


print("\nEnter weights for Hidden Layer 2")

w31 = float(input("H1 -> H3: "))
w32 = float(input("H2 -> H3: "))
w41 = float(input("H1 -> H4: "))
w42 = float(input("H2 -> H4: "))

W2 = np.array([
    [w31, w41],
    [w32, w42]
])


print("\nEnter biases for Hidden Layer 2")

b2_1 = float(input("Bias H3: "))
b2_2 = float(input("Bias H4: "))

b2 = np.array([[b2_1, b2_2]])


print("\nEnter weights for Output Layer")

w51 = float(input("H3 -> Output: "))
w52 = float(input("H4 -> Output: "))

W3 = np.array([
    [w51],
    [w52]
])


b3_value = float(input("Enter output bias: "))
b3 = np.array([[b3_value]])


for epoch in range(epochs):

    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)

    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    z3 = np.dot(a2, W3) + b3
    output = sigmoid(z3)

    loss = mse_loss(Y, output)

    if epoch == 0:
        print("\n========== FIRST ITERATION ==========")
        print("Epoch:", epoch + 1)
        print("Z1 =", z1)
        print("A1 =", a1)
        print("Z2 =", z2)
        print("A2 =", a2)
        print("Z3 =", z3)
        print("Output =", output)
        print("MSE Loss =", loss)

    delta3 = 2 * (output - Y) * sigmoid_derivative(output)

    delta2 = np.dot(delta3, W3.T) * sigmoid_derivative(a2)

    delta1 = np.dot(delta2, W2.T) * sigmoid_derivative(a1)

    W3 -= np.dot(a2.T, delta3) * lr
    b3 -= delta3 * lr

    W2 -= np.dot(a1.T, delta2) * lr
    b2 -= delta2 * lr

    W1 -= np.dot(X.T, delta1) * lr
    b1 -= delta1 * lr

    if epoch == epochs - 1:
        print("\n========== LAST ITERATION ==========")
        print("Epoch:", epoch + 1)
        print("Z1 =", z1)
        print("A1 =", a1)
        print("Z2 =", z2)
        print("A2 =", a2)
        print("Z3 =", z3)
        print("Output =", output)
        print("MSE Loss =", loss)

    if (epoch + 1) % 100 == 0:
        print("Epoch:", epoch + 1, "MSE Loss:", loss)


print("\n========== FINAL RESULT ==========")

print("Input:", X)
print("Target:", Y)
print("Predicted Output:", output)
print("Final MSE Loss:", mse_loss(Y, output))

if output[0][0] >= 0.5:
    print("Predicted Class: 1")
else:
    print("Predicted Class: 0")


print("\nUpdated Weights:")

print("W1 =\n", W1)
print("W2 =\n", W2)
print("W3 =\n", W3)


print("\nUpdated Biases:")

print("B1 =", b1)
print("B2 =", b2)
print("B3 =", b3)
```
