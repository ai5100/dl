```python
import math
import random

# Sigmoid Activation Function
def sigmoid(x):
    return 1 / (1 + math.exp(-x))

# Convert sigmoid output to binary
def predict(y):
    if y >= 0.5:
        return 1
    return 0

# XOR Dataset
inputs = [
    (0, 0),
    (0, 1),
    (1, 0),
    (1, 1)
]

targets = [0, 1, 1, 0]

print("========== MLP XOR : Forward Propagation ==========\n")

choice = input("Do you want to enter weights manually? (y/n): ")

if choice.lower() == 'y':

    print("\nInput -> Hidden Layer")

    w11 = float(input("Weight I1 -> H1 : "))
    w21 = float(input("Weight I2 -> H1 : "))
    w12 = float(input("Weight I1 -> H2 : "))
    w22 = float(input("Weight I2 -> H2 : "))

    b1 = float(input("Bias of H1 : "))
    b2 = float(input("Bias of H2 : "))

    print("\nHidden -> Output Layer")

    v1 = float(input("Weight H1 -> O : "))
    v2 = float(input("Weight H2 -> O : "))

    bo = float(input("Output Bias : "))

else:
    w11 = w21 = w12 = w22 = 0
    b1 = b2 = 0
    v1 = v2 = 0
    bo = 0


# ---------- Check Manual Weights ----------

def test_network():

    outputs = []

    for x1, x2 in inputs:

        h1 = sigmoid(x1 * w11 + x2 * w21 + b1)
        h2 = sigmoid(x1 * w12 + x2 * w22 + b2)

        out = sigmoid(h1 * v1 + h2 * v2 + bo)

        outputs.append(predict(out))

    return outputs


outputs = test_network()

# If wrong output, generate random weights
if outputs != targets:

    print("\nEntered weights do not produce XOR.")
    print("Searching random weights...\n")

    found = False

    while not found:

        w11 = random.uniform(-10, 10)
        w21 = random.uniform(-10, 10)

        w12 = random.uniform(-10, 10)
        w22 = random.uniform(-10, 10)

        b1 = random.uniform(-10, 10)
        b2 = random.uniform(-10, 10)

        v1 = random.uniform(-10, 10)
        v2 = random.uniform(-10, 10)

        bo = random.uniform(-10, 10)

        outputs = test_network()

        if outputs == targets:
            found = True

print("Weights Used\n")

print("w11 =", round(w11,3))
print("w21 =", round(w21,3))
print("w12 =", round(w12,3))
print("w22 =", round(w22,3))

print("b1 =", round(b1,3))
print("b2 =", round(b2,3))

print("v1 =", round(v1,3))
print("v2 =", round(v2,3))

print("bo =", round(bo,3))

print("\nForward Propagation Output\n")

print("Input\tExpected\tPredicted")

for i in range(4):

    x1, x2 = inputs[i]

    h1 = sigmoid(x1 * w11 + x2 * w21 + b1)
    h2 = sigmoid(x1 * w12 + x2 * w22 + b2)

    out = sigmoid(h1 * v1 + h2 * v2 + bo)

    pred = predict(out)

    print(f"{x1} {x2}\t   {targets[i]}\t\t   {pred}")




import random
import math


X = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

Y = [0, 1, 1, 0]


def sigmoid(x):
    return 1 / (1 + math.exp(-x))


def dsigmoid(y):
    return y * (1 - y)

print("========== MLP XOR using Backpropagation ==========")

choice = input("Enter weights manually? (y/n): ")

if choice.lower() == 'y':

    print("\n--- Input Layer to Hidden Layer ---")
    w11 = float(input("Weight I1 -> H1 : "))
    w21 = float(input("Weight I2 -> H1 : "))
    w12 = float(input("Weight I1 -> H2 : "))
    w22 = float(input("Weight I2 -> H2 : "))

    print("\n--- Hidden Layer Biases ---")
    b1 = float(input("Bias of H1 : "))
    b2 = float(input("Bias of H2 : "))

    print("\n--- Hidden Layer to Output Layer ---")
    v1 = float(input("Weight H1 -> O : "))
    v2 = float(input("Weight H2 -> O : "))

    print("\n--- Output Bias ---")
    bo = float(input("Bias of Output Neuron : "))

else:
    print("\nInitializing weights randomly...\n")

    w11 = random.uniform(-1, 1)
    w21 = random.uniform(-1, 1)

    w12 = random.uniform(-1, 1)
    w22 = random.uniform(-1, 1)

    v1 = random.uniform(-1, 1)
    v2 = random.uniform(-1, 1)

    b1 = random.uniform(-1, 1)
    b2 = random.uniform(-1, 1)
    bo = random.uniform(-1, 1)


lr = 0.5
epochs = 10000


for epoch in range(epochs):

    total_error = 0

    for i in range(len(X)):

        x1, x2 = X[i]
        target = Y[i]


        h1 = sigmoid(x1 * w11 + x2 * w21 + b1)
        h2 = sigmoid(x1 * w12 + x2 * w22 + b2)

        out = sigmoid(h1 * v1 + h2 * v2 + bo)


        error = target - out
        total_error += error ** 2


        delta_out = error * dsigmoid(out)

        delta_h1 = dsigmoid(h1) * delta_out * v1
        delta_h2 = dsigmoid(h2) * delta_out * v2


        v1 += lr * delta_out * h1
        v2 += lr * delta_out * h2
        bo += lr * delta_out


        w11 += lr * delta_h1 * x1
        w21 += lr * delta_h1 * x2
        b1 += lr * delta_h1

        w12 += lr * delta_h2 * x1
        w22 += lr * delta_h2 * x2
        b2 += lr * delta_h2

    if epoch % 1000 == 0:
        print(f"Epoch {epoch:5d}  Error = {total_error:.6f}")


print("\n========== Trained Weights ==========")

print(f"w11 = {w11:.4f}")
print(f"w21 = {w21:.4f}")
print(f"w12 = {w12:.4f}")
print(f"w22 = {w22:.4f}")

print(f"b1  = {b1:.4f}")
print(f"b2  = {b2:.4f}")

print(f"v1  = {v1:.4f}")
print(f"v2  = {v2:.4f}")

print(f"bo  = {bo:.4f}")


print("\n========== Final Outputs ==========")
print("Input\tTarget\tOutput\tPrediction")

for i in range(len(X)):

    x1, x2 = X[i]

    h1 = sigmoid(x1 * w11 + x2 * w21 + b1)
    h2 = sigmoid(x1 * w12 + x2 * w22 + b2)

    out = sigmoid(h1 * v1 + h2 * v2 + bo)

    pred = 1 if out >= 0.5 else 0

    print(f"{X[i]}\t{Y[i]}\t{out:.4f}\t{pred}")
```
